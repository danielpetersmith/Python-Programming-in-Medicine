## List comprehension syntax


List comprehensions use the following syntax in Python:

`my_list = [i for i in my_iterable if my_condition]`

Where: 
* *my_iterable* is something like a string, list, or tuple (i.e. something that can be *iterated* through).
* *my_condition* is some condition that you check that returns True or False (e.g. num >= 3).
* *i* is included in the list if this *my_condition* is True

'i' here is a 'dummy variable' and is exactly the same as the one you will know from a typical *for* loop ("for i in ...")
<br>
Once the list comprehension has been run we assign its output to a varaible, in this case the variable is *'my_list'*


**Let's look at a more concrete example:**

````python
# try running this code

my_sequence = (5, 3, -2, 1, -7, 9, 6)
negatives = [i for i in my_sequence if i < 0]
print(negatives)
````

In the above code, we start with a tuple called *my_sequence* and want to make a list of only the negative numbers in *my_sequence*. <br>
To do this, the list comprehension:
* iterates through my_sequence (finds the next element of *my_sequence* and assigns its value to 'i')
* checks if that value ( i ) is less than 0
* if so, that value is included in the list; conversely, if i >= 0, we do **not** include it in the list
* once it reaches the end of *my_sequence*, the list is assigned to *negatives*.
<br>

The code above behaves exactly the same as the *for* loop below, which is hopefully more familiar:

````python
my_sequence = (5, 3, -2, 1, -7, 9, 6)
negatives = []
for i in my_sequence:
    if i < 0:
        negatives.append(i)
print(negatives)
````
